```python
def NTT(x, phi, mod):
    N = len(x)
    if N == 1:
        return x

    E = NTT(x[0::2], pow(phi, 2, mod), mod)
    O = NTT(x[1::2], pow(phi, 2, mod), mod)

    X = [0] * N
    w = 1

    for k in range(N // 2):
        t = (w * O[k]) % mod
        X[k] = (E[k] + t) % mod
        X[k + N//2] = (E[k] - t) % mod
        w = (w * phi) % mod

    return X
```

```python
def inverse_NTT(a, root, mod):
    n = len(a)

    # Inverse root
    inv_root = pow(root, mod - 2, mod)

    # Apply forward NTT using inverse root
    res = NTT(a, inv_root, mod)

    # Multiply by n^{-1}
    inv_n = pow(n, mod - 2, mod)
    for i in range(n):
        res[i] = (res[i] * inv_n) % mod

    return res
```
