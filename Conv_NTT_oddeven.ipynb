Matrix Multiplication

```python
## Toeplitz constuction
def toep_construction(seed, row, clm):
    toep = np.zeros((row, clm))
    for i in range(0,row):
        for j in range(0,clm):
            toep[i,j] = seed[row - 1 + j - i]
    return toep
```

```python
# for hashing input = coloumns and output = rows
# generate toep
toep = toep_construction(seed[::-1], output_len, input_len)
print(toep)
```

```python
# Matrix mul
output_matrix_mul = np.dot(toep, input) 
print(output_matrix_mul)
```

NTT

```python
def NTT(x, phi, mod):
    N = len(x)
    if N == 1:
        return x

    E = NTT(x[0::2], pow(phi, 2, mod), mod)
    O = NTT(x[1::2], pow(phi, 2, mod), mod)

    X = [0] * N
    w = 1

    for k in range(N // 2):
        t = (w * O[k]) % mod
        X[k] = (E[k] + t) % mod
        X[k + N//2] = (E[k] - t) % mod
        w = (w * phi) % mod

    return X
```

```python
def inverse_NTT(a, root, mod):
    n = len(a)

    # Inverse root
    inv_root = pow(root, mod - 2, mod)

    # Apply forward NTT using inverse root
    res = NTT(a, inv_root, mod)

    # Multiply by n^{-1}
    inv_n = pow(n, mod - 2, mod)
    for i in range(n):
        res[i] = (res[i] * inv_n) % mod

    return res
```

```python
import tkinter as tk
import math

class IceStudyTimer:
    def __init__(self, root):
        self.root = root
        self.root.title("🧊 Save Your Beer")
        self.root.geometry("420x360")
        self.root.configure(bg="#0f172a")

        # Core state
        self.total_time = 1800
        self.remaining = self.total_time
        self.running = False
        self.ice_level = 100
        self.phase = 0

        # ===== UI =====
        self.title = tk.Label(root, text="🧊 SAVE THE BEER",
                              font=("Helvetica", 16, "bold"), fg="#38bdf8", bg="#0f172a")
        self.title.pack(pady=8)

        self.timer = tk.Label(root, text=self.format_time(self.remaining),
                              font=("Helvetica", 36, "bold"), fg="#e0f2fe", bg="#0f172a")
        self.timer.pack(pady=5)

        # Canvas
        self.canvas = tk.Canvas(root, width=320, height=180, bg="#111827", highlightthickness=0)
        self.canvas.pack(pady=10)

        # ===== 3D ICE BLOCK (fake perspective) =====
        # front face
        self.front = self.canvas.create_rectangle(60, 60, 260, 160, fill="#7dd3fc", outline="")
        # top face
        self.top = self.canvas.create_polygon(60,60, 90,40, 290,40, 260,60,
                                              fill="#bae6fd", outline="")
        # side face
        self.side = self.canvas.create_polygon(260,60, 290,40, 290,140, 260,160,
                                               fill="#38bdf8", outline="")

        # Beer (simple but centered)
        self.beer_body = self.canvas.create_rectangle(145, 100, 175, 135, fill="#facc15", outline="")
        self.beer_foam = self.canvas.create_oval(145, 85, 175, 105, fill="#ffffff", outline="")

        # Status
        self.status = tk.Label(root, text="Ice Remaining: 100%",
                               font=("Helvetica", 11), fg="#cbd5f5", bg="#0f172a")
        self.status.pack()

        # Buttons
        frame = tk.Frame(root, bg="#0f172a")
        frame.pack(pady=10)

        tk.Button(frame, text="Start", command=self.start, width=8).grid(row=0, column=0, padx=5)
        tk.Button(frame, text="Pause", command=self.pause, width=8).grid(row=0, column=1, padx=5)
        tk.Button(frame, text="Reset", command=self.reset, width=8).grid(row=0, column=2, padx=5)

        self.update_timer()
        self.animate()

    def format_time(self, s):
        return f"{s//60:02}:{s%60:02}"

    def update_timer(self):
        if self.running and self.remaining > 0:
            self.remaining -= 1
            self.ice_level = max(self.ice_level - (100/self.total_time), 0)

            self.timer.config(text=self.format_time(self.remaining))
            self.status.config(text=f"Ice Remaining: {self.ice_level:.1f}%")

        self.root.after(1000, self.update_timer)

    def animate(self):
        self.phase += 0.1

        # ===== MELTING EFFECT (3D shrink) =====
        h = 100 * (self.ice_level/100)
        top_y = 160 - h

        # front face shrinks
        self.canvas.coords(self.front, 60, top_y, 260, 160)

        # top face follows
        self.canvas.coords(self.top,
            60, top_y,
            90, top_y-20,
            290, top_y-20,
            260, top_y
        )

        # side face follows
        self.canvas.coords(self.side,
            260, top_y,
            290, top_y-20,
            290, 140,
            260, 160
        )

        # subtle lighting change (fake 3D shading)
        intensity = int(180 + 60*math.sin(self.phase))
        front_color = f"#7d{intensity:02x}fc"
        self.canvas.itemconfig(self.front, fill=front_color)

        # beer slight float (inside ice illusion)
        dy = 1.5 * math.sin(self.phase)
        self.canvas.move(self.beer_body, 0, dy)
        self.canvas.move(self.beer_foam, 0, dy)
        self.canvas.move(self.beer_body, 0, -dy)
        self.canvas.move(self.beer_foam, 0, -dy)

        # panic effect
        if self.ice_level < 10:
            self.root.geometry(f"420x360+{int(3*math.sin(self.phase))}+{int(3*math.cos(self.phase))}")

        # timer glow
        glow = int(200 + 55*math.sin(self.phase))
        self.timer.config(fg=f"#00{glow:02x}ff")

        self.root.after(50, self.animate)

    def start(self): self.running = True
    def pause(self): self.running = False

    def reset(self):
        self.running = False
        self.remaining = self.total_time
        self.ice_level = 100
        self.timer.config(text=self.format_time(self.remaining))
        self.status.config(text="Ice Remaining: 100%")


if __name__ == "__main__":
    root = tk.Tk()
    app = IceStudyTimer(root)
    root.mainloop()
```
