```python
def fft(x):
    N = len(x)
    if N == 1:
        return x

    # FFT of even and odd indices
    E = fft(x[0::2])
    O = fft(x[1::2])

    # Twiddle factors
    W = np.exp(-2j * np.pi * np.arange(N) / N)

    # Combine
    X = np.zeros(N, dtype=complex)
    for k in range(N // 2):
        X[k]         = E[k] + W[k] * O[k]
        X[k + N//2]  = E[k] - W[k] * O[k]

    return X

        

```

```python
import numpy as np

# Coefficients of the polynomial
a = np.array([1, 2, 3, 4], dtype=float)

# Zero-pad to avoid circular convolution
n = len(a)
size = 2 * n - 1

A = np.fft.fft(a, size)
C = A * A
c = np.fft.ifft(C)
print(c)
# Remove tiny numerical errors
c = np.round(np.real(c)).astype(int)

print("Result coefficients:")
print(c)
```
